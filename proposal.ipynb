# Motivation & Problem 
- I want to find out where in Seattle people can take dance classes and want to know which styles, and times are easy to find. As a dancer, I've had to check many studio websites just to find one class and no website shows everything. This analysis would show how classes are spread across different areas and would help dancers find options near them. I hope to learn which neighborhoods typically have the most classes and which styles cluster in certain areas.

# Data
- I plan to collect this data by hand from public studio websites. I will cover around 15 to 20 various studios in the greater Seattle area. Each row will have one class per week, and the columns will be studio, neighborhood, style, day, start time, level, source link, and date collected. I will link every studio page I use, and record the snapshot date as schedules can change.
- This data fits my question because it is small, local, and something that I can realistically finish in this term. I will check all the studio's terms to not about copying or reusing before I collect.
- For ethics, I will be collecting data based on businesses, not people. I will ensure to leave out instructor's names and personal details. I will also make sure to pick out studios from different neighborhoods and not just the ones I know to reduce bias on who gets counted.

# Unknowns & Dependencies
- Studios may change their schedules before I finish
- Some sites hid schedules behind booking tools, making it hard to record
- A site's terms might limit how I share the schedule
- My sample might miss smaller studios that don't have websites

# Overview & Research Questions
Seattle dance classes are posted on many separate studio and company websites. A dancer has to already know a studio to find its schedule. This is hardest for beginners, who don't yet know where to look. It also makes it hard for studios, who must reach out to dancers one by one. This study will look at what the scattered listings actually contain. It asks whether classes are easy to find, especially for beginners. I hope the results show where the information is missing. They could guide a future tool that brings listings together to one platform. This study will describe class availabilty in my sample, but does not measure enrollement or how studios reach dancers.

Questions:
-  Which Seattle neighborhoods offer the most classes in my sample?
- Are there certain days and times that tend to offer more beginner friendly classes?
- What days and times offer the most classes?

Hypothesis: 
- I expect class levels to be hard to identify as many sites don't list levels or have it as an "open" class.
- I expect classes to cluster in a few central areas.

# Background & Related Work
- Researchers have studied where arts and recreation programs are located in cities. Fatooh (2022) maps Seattle arts organizations by decade, and found that art activity is densest in Uptown, with smaller clusters in Pioneer Square, Capitol Hill, and the Central District. She also noted that the map may favor better-resourced organizations. 
- Other Studies show access can differ by neighborhood. Powell et al. (2006) compared neighborhood demographics with the number of dance facilities and other fitness settings across the U.S.They found that there were fewer dance facilites in lower-income areas and in areas with higher shares of Black, Hispanic, and other minority residents. An Active Living Research study found that cities with less fiscal capacity and more non-white residents had less access to organized recreation programs. This is why I am planning on comparing neighborhoods, and my study will not use income or race data, but only describe where classes are located. 
- I found no research on how dancers search for classes, and my study will not answer that as it will only describe what the listings contain.

# Methodology
- Data Collection:
    - I will collect class schedules by hand from public forums of about 15 to 20 Seattle dance studios.
    - I will pick studios from different neighborhoods.
    - I will record all classes on one snapshot date, and each row is one weekly class. 
    - For each class, I will record the studio, neighborbood, style, day, time, level, source link, and the date I collected the data. If the site does not list a level, I will write "not listed"
- Methods:
    - Question 1 (neighborhoods): count classes and studios in each neighborhood.
    - Question 2 (beginner-friendly days and times): count classes labeled beginner by day and by time of day. Then compare them will all other classes.
    - Question 3 (most classes offered by day and time): count all classes by day and time of day.
- Why my methods fit:
    - My sample is small enough that I will be able to collect it myself. 
    - Simple counts are clear about what the data shows.
- How I will present my findings:
    - A bar chart of classes per neighborhood.
    - A bar chart of classes per style, and a table of which neighborhoods offer each style.
    - A bar chart of beginner, and "not listed: classes.
    - A table of classes by day and time of day.
- Limits:
    - My sample will not include every studio or company in Seattle and may miss smaller studios with no website.
    - Schedules can change, so some of the data I collect could change.
    - My results will describe my sample, not all of Seattle.
