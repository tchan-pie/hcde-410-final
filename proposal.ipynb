# Motivation & Problem 
- I want to find out where in Seattle people can take dance classes and want to know which styles, and times are easy to find. As a dancer, I've had to check many studio websites just to find one class and no website shows everything. This analysis would show how classes are spread across different areas and would help dancers find options near them. I hope to learn which neighborhoods typically have the most classes and which styles cluster in certain areas.

# Data
- I plan to collect this data by hand from public studio websites. I will cover around 15 to 20 various studios in the greater Seattle area. Each row will have one class per week, and the columns will be studio, neighborhood, style, day, start time, level, source link, and date collected. I will link every studio page I use, and record the snapshot date as schedules can change.
- This data fits my question because it is small, local, and something that I can realistically finish in this term. I will check all the studio's terms to not about copying or reusing before I collect.
- For ethics, I will be collecting data based on businesses, not people. I will ensure to leave out instructor's names and personal details. I will also make sure to pick out studios from different neighborhoods and not just the ones I know to reduce bias on who gets counted.

# Unknowns & Dependencies
- Studios may change their schedules before I finish
- Some sites hid schedules behind booking tools, making it hard to record
- A site's terms might limit how I share the schedule
- My sample might miss smaller studios that don't have websites
